# Egenværdier og egenvektorer og løsning af differentialligningssystemer med SymPy
Denne notebook demonstrerer, hvordan man finder egenværdier, egenvektorer og løser differentialligninger med med SymPy.

```python
import sympy as sp

sp.init_printing()  # gør at matematiske udtryk skrives pænt

# Definer en matrix
A = sp.Matrix([[3, -2], [5, -3]])
A
```

## Beregning af egenværdier

```python
eigenvals = A.eigenvals()
eigenvals
```

## Beregning af egenvektorer

```python
eigenvects = A.eigenvects()
eigenvects
```

# Systemer af differentialligninger i SymPy


```python
t = sp.symbols('t')
x1 = sp.Function('x1')
x2 = sp.Function('x2')
```

## System:

```python
eq1 = sp.Eq(sp.diff(x1(t), t), -x1(t) + 2*x2(t))
eq2 = sp.Eq(sp.diff(x2(t), t), -2*x2(t))
display(eq1)
display(eq2)
```

## Fuldstændig løsning

```python
sol = sp.dsolve((eq1, eq2))
display(sol)
```

## Løsning med begyndelsesværdier

```python
ics = {x1(0):3, x2(0):4}
sol_ics = sp.dsolve((eq1, eq2), ics=ics)
display(sol_ics)
```
